# Airbnb Listing Market Analysis
1. Business problem and project objective
Business problem

Airbnb hosts and market researchers need to understand how listing prices vary across room types and neighbourhoods, and whether listing characteristics such as availability, reviews, and accommodation capacity are associated with pricing.

Without this analysis, hosts may struggle to compare their listing with similar properties, identify areas with different price levels, or understand which listing attributes are worth investigating further.

# Project objective
- Explore Airbnb listing data and assess its quality.
- Compare prices across room types and neighbourhood groups.
- Investigate relationships between price, reviews, availability, and accommodation features.
- Identify patterns that could help hosts make more informed pricing and listing decisions.

```python
# 1. Import important libraries which we are going to use

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# 2. Read CSV File

df=pd.read_csv(r'C:\Akshay\Data Analytics\Projects Practice\Python EDA\NYC Airbnb\datasets.csv')
df.head(10)
```

```python
# 3. DATA INSPECTION  

# 1. Dataset size
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 2. Missing values
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
}).sort_values("Missing_Percentage", ascending=False)

display(missing_summary)

# 3. Duplicate checks
print("Duplicate rows:", df.duplicated().sum())

if "id" in df.columns:
    print("Duplicate listing IDs:", df["id"].duplicated().sum())

# 4. Numeric validation
for col in ["price", "minimum_nights",
             "availability_365", "number_of_reviews"]:
    if col in df.columns:
        print(f"\n{col}")
        print("Negative values:", (df[col] < 0).sum())
```

# Data Cleaning

```python
# 4. Drop duplicates and null values

df.isnull().sum()
df.dropna(inplace=True)
df.info()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
# 5. Change Data Types (Type casting)

df['id']=df['id'].astype('object')
df['host_id']=df['host_id'].astype('object')
df['name']=df['name'].astype('object')
df['host_name']=df['host_name'].astype('object')
df['neighbourhood_group']=df['neighbourhood_group'].astype('object')
df['neighbourhood']=df['neighbourhood'].astype('object')
df['room_type']=df['room_type'].astype('object')
df['last_review']=df['last_review'].astype('object')
df['license']=df['license'].astype('object')
df['rating']=df['rating'].astype('object')
df['bedrooms']=df['bedrooms'].astype('object')
df['baths']=df['baths'].astype('object')
df.dtypes
```

# Univariate Analysis

```python
df.describe()
```

```python
price_summary = df.groupby("room_type")["price"].agg(
    Listing_Count="count",
    Median_Price="median",
    Mean_Price="mean",
    Min_Price="min",
    Max_Price="max"
)

display(price_summary.sort_values("Median_Price"))

# Percentiles
percentiles = df["price"].quantile(
    [0.25, 0.50, 0.75, 0.90, 0.95]
)

display(percentiles)
```

```python
# Identyfying Outliers in price
data= df[df['price']<1500]
sns.boxplot(x='price', data=data)
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=data, x='price', bins=50)
plt.title('Price Distribution')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show
```

 Outcome :  The price distribution is right-skewed: the median listing price is 125, while the 95th percentile is 429.85. This indicates that a relatively small portion of listings are priced substantially higher than the typical listing.

```python
plt.figure(figsize=(6, 3))
sns.histplot(data=df, x='availability_365')
plt.title('availability_365 Distribuition')
plt.ylabel("Frequency")
plt.show()
```

```python
df.dtypes
```

```python
# price per neighbourhood group
df.groupby(by='neighbourhood_group')['price'].mean()
```

```python
# price per bed
df['price per bed']= df['price']/df['beds']
df.groupby(by='neighbourhood_group')['price per bed'].mean()
```

 Manhattan has the highest mean price per bed at approximately 151.93, followed by Brooklyn at 128.36. Staten Island has the 
lowest mean price per bed at approximately 67.73.

This provides a capacity-adjusted comparison of listing prices across neighbourhood groups.

```python
df["availability_segment"] = pd.cut(
    df["availability_365"],
    bins=[-1, 90, 180, 270, 365],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Very High"
    ]
)

availability_analysis = df.groupby(
    "availability_segment",
    observed=True
).agg(
    Listing_Count=("id", "nunique"),
    Median_Price=("price", "median"),
    Median_Reviews=("number_of_reviews", "median"),
    Median_Availability=("availability_365", "median")
)

display(availability_analysis)
```

# Overall Conclusion

Overall Conclusion

This exploratory data analysis examined Airbnb listing data to understand how listing prices vary across room types and neighbourhood groups, and how pricing relates to listing characteristics such as availability, reviews, and accommodation capacity.

The analysis identified several important patterns:

1. Accommodation type: Entire homes/apartments have a median listing price of 160, compared with 80 for private rooms. Hotel rooms have the highest median and mean prices among the room types in the dataset, although the hotel-room category contains fewer listings.

2. Geographical pricing: Manhattan has the highest average listing price at approximately 227.48, followed by Brooklyn at 187.20. Bronx and Staten Island have lower average listing prices.

3. Price distribution: The listing-price distribution is right-skewed, with a median of 125 and a 95th-percentile price of 429.85. This highlights the importance of examining outliers and using median prices alongside averages.

4. Price relative to accommodation capacity: Manhattan has the highest mean price per bed at approximately 151.93, while Staten Island has the lowest at approximately 67.73.

Overall, the findings indicate that room type and geographical location are important dimensions for comparing Airbnb listing prices. The analysis provides a foundation for more detailed pricing research and for helping hosts benchmark their listings against similar properties.
